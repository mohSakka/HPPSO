```python
import autograd.numpy as np
from autograd import grad
import numpy as np_orig

# Define benchmark functions and their gradients

######## Rastrigin #################
def rastrigin_function(x):
    x = np.atleast_1d(x)  # Ensure x is an array
    return 10 * len(x) + np.sum(np.array([xi**2 - 10 * np.cos(2 * np.pi * xi) for xi in x]))

####### Ackley ##############################
def ackley_function(x):
    x = np.atleast_1d(x)  # Ensure x is an array
    part1 = -0.2 * np.sqrt(np.mean(x**2))
    part2 = np.mean(np.cos(2 * np.pi * x))
    return -20 * np.exp(part1) - np.exp(part2) + 20 + np.e

######### Rosenbrock ##########################
def rosenbrock_function(x):
    x = np.atleast_1d(x)  # Ensure x is an array
    return np.sum(np.array([100 * (x[i+1] - xi**2)**2 + (1 - xi)**2 for i, xi in enumerate(x[:-1])]))

########### Beale ##################################
def beale_function(x):
    y = x[1]
    x = x[0]
    return (1.5 - x + x * y)**2 + (2.25 - x + x * y**2)**2 + (2.625 - x + x * y**3)**2

################# GOLDSTEIN-PRICE #################
def goldstein_price_function(x):
    y = x[1]
    x = x[0]
    term1 = 1 + (x + y + 1)**2 * (19 - 14*x + 3*x**2 - 14*y + 6*x*y + 3*y**2)
    term2 = 30 + (2*x - 3*y)**2 * (18 - 32*x + 12*x**2 + 48*y - 36*x*y + 27*y**2)
    return term1 * term2

############## Bukin Function N.6 ##########
def bukin_function_n6(x):
    y = x[1]
    x = x[0]
    return 100 * np.sqrt(np.abs(y - 0.01 * x**2)) + 0.01 * np.abs(x + 10)

########## Levy Function N.13 ###################
def levy_function_n13(x):
    y = x[1]
    x = x[0]
    w1 = 1 + (x - 1) / 4
    w2 = 1 + (y - 1) / 4
    term1 = np.sin(np.pi * w1)**2
    term2 = (w1 - 1)**2 * (1 + 10 * np.sin(np.pi * w1 + 1)**2)
    term3 = (w2 - 1)**2 * (1 + np.sin(2 * np.pi * w2)**2)
    return term1 + term2 + term3

############### Himmelblau ###################
def himmelblau_function(x):
    y = x[1]
    x = x[0]
    return (x**2 + y - 11)**2 + (x + y**2 - 7)**2

############## Three-Hump Camel ###############
def three_hump_camel_function(x):
    y = x[1]
    x = x[0]
    return 2 * x**2 - 1.05 * x**4 + x**6 / 6 + x * y + y**2

########## Griewank ################################
def griewank_function(x):
    x = np.atleast_1d(x)  # Ensure x is an array
    part1 = np.sum(np.array([xi ** 2 / 4000 for xi in x]))
    part2 = np.prod(np.array([np.cos(xi / np.sqrt(i+1)) for i, xi in enumerate(x)]))
    return part1 - part2 + 1

######## Cross-In-Tray ########################
def cross_in_tray(x):
    y = x[1]
    x = x[0]
    pi = np.pi
    sin_x = np.sin(x)
    sin_y = np.sin(y)
    exp_term = np.exp(np.abs(100 - np.sqrt(x**2 + y**2)) / pi)

    return -0.0001 * (np.abs(sin_x * sin_y * exp_term) + 1) ** 0.1

########### EggHolder ################
def eggholder_function(x):
    y = x[1]
    x = x[0]
    return -(y + 47) * np.sin(np.sqrt(np.abs(x / 2 + (y + 47)))) - x * np.sin(np.sqrt(np.abs(x - (y + 47))))

########### Holder Table ################
def holder_table_function(x):
    y = x[1]
    x = x[0]
    return -np.abs(np.sin(x) * np.cos(y) * np.exp(np.abs(1 - np.sqrt(x**2 + y**2) / np.pi)))


# 1. Michalewicz Function
def michalewicz(x):
    m = 10
    return -np.sum(np.sin(x) * (np.sin(((np.arange(1, len(x) + 1) * x**2) / np.pi)))**(2 * m))

# 2. Alpine Function
def alpine(x):
    return np.sum(np.abs(x * np.sin(x) + 0.1 * x))

# 3. Schaffer Function
def schaffer(x):
    x1, x2 = x
    return 0.5 + (np.sin(x1**2 - x2**2)**2 - 0.5) / ((1 + 0.001 * (x1**2 + x2**2))**2)

# 4. Easom Function
def easom(x):
    x1, x2 = x
    return -np.cos(x1) * np.cos(x2) * np.exp(-((x1 - np.pi)**2 + (x2 - np.pi)**2))

# 5. Shubert Function
def shubert(x):
    x1, x2 = x
    sum1 = np.sum([i * np.cos((i + 1) * x1 + i) for i in range(1, 6)])
    sum2 = np.sum([i * np.cos((i + 1) * x2 + i) for i in range(1, 6)])
    return sum1 * sum2

# 6. Shekel Function (Shekel 4.5)
def shekel(x):
    m = 5
    C = np.array([
        [4.0, 4.0, 4.0, 4.0],
        [1.0, 1.0, 1.0, 1.0],
        [8.0, 8.0, 8.0, 8.0],
        [6.0, 6.0, 6.0, 6.0],
        [3.0, 7.0, 3.0, 7.0]
    ])
    beta = np.array([0.1, 0.2, 0.2, 0.4, 0.4])
    outer = 0
    for i in range(m):
        inner = np.sum((x - C[i, :])**2)
        outer += 1 / (inner + beta[i])
    return -outer

# 7. Trid Function
def trid(x):
    n = len(x)
    sum1 = np.sum((x - 1)**2)
    sum2 = np.sum(x[1:] * x[:-1])
    return sum1 - sum2

# 8. Zettl Function
def zettl(x):
    x1, x2 = x
    return (x1**2 + x2**2 - 2*x1)**2 + 0.25 * x1

# 9. Leon Function
def leon(x):
    x1, x2 = x
    return 100 * (x2 - x1**3)**2 + (1 - x1)**2

def drop_wave(x):
    x1, x2 = x
    numerator = 1 + np.cos(12 * np.sqrt(x1**2 + x2**2))
    denominator = 0.5 * (x1**2 + x2**2) + 2
    return -numerator / denominator

def langermann(x):
    m = 5
    a = np.array([3, 5, 2, 1, 7])
    b = np.array([5, 2, 1, 4, 9])
    c = np.array([1, 2, 5, 2, 3])

    sum_term = 0
    for i in range(m):
        sum_term += c[i] * np.exp(-1/np.pi * ((x[0] - a[i])**2 + (x[1] - b[i])**2)) * np.cos(np.pi * ((x[0] - a[i])**2 + (x[1] - b[i])**2))

    return -sum_term

def schwefel(x):
    n = len(x)
    return 418.9829 * n - np.sum(x * np.sin(np.sqrt(np.abs(x))))

def schaffer_function_n4(x):
    """
    Computes the value of the Schaffer Function N. 4 for a given input x.

    Parameters:
    x (list or np.array): A list or array containing two values, x1 and x2.

    Returns:
    float: The computed Schaffer Function N. 4 value.
    """
    # Decompose the input x into x1 and x2
    x1, x2 = x[0], x[1]

    # Compute the numerator and denominator as per the function's formula
    numerator = np.cos(np.sin(np.abs(x1**2 - x2**2)))**2 - 0.5
    denominator = (1 + 0.001 * (x1**2 + x2**2))**2

    # Calculate the final result
    result = 0.5 + numerator / denominator
    return result

# Example usage:
x = np.array([2.20, 1.57])
print("Michalewicz function value:", michalewicz(x))
print("Alpine function value:", alpine(x))
print("Schaffer function value:", schaffer(x))
print("Easom function value:", easom(x))
print("Shubert function value:", shubert(x))
print("Shekel function value:", shekel(np.array([1.0, 1.0, 1.0, 1.0])))
print("Trid function value:", trid(x))
print("Zettl function value:", zettl(x))
print("Leon function value:", leon(x))

# Example usage
if __name__ == "__main__":
    # Create Autograd arrays
    x = np.array([0, 0])
    print("Rastrigin:", rastrigin_function(x))
    print("Ackley:", ackley_function(x))
    print("Rosenbrock:", rosenbrock_function(x))
    print("Beale:", beale_function(x))
    print("Goldstein-Price:", goldstein_price_function(x))
    print("Bukin N.6:", bukin_function_n6(x))
    print("Levy N.13:", levy_function_n13(x))
    print("Himmelblau:", himmelblau_function(x))
    print("Three-hump camel:", three_hump_camel_function(x))
    print("Griewank:", griewank_function(x))
    print("Cross-in-tray:", cross_in_tray(x))
    print("Eggholder:", eggholder_function(x))
    print("Holder table:", holder_table_function(x))
```

```python
class GradientDescent:
    def __init__(self, cost_function, learning_rate=0.01, tol=1e-4, max_epochs=100, seed=10, init_range=[-5.12,5.12]):
        self.cost_function = cost_function
        self.gradient_function = grad(self.cost_function)
        self.learning_rate = learning_rate
        self.tol = tol # convergence tolerance
        self.max_epochs = max_epochs
        self.seed = seed
        self.init_range = init_range

    def run(self, dimensions):
        np_orig.random.seed(self.seed)
        weights = np_orig.random.uniform(low=self.init_range[0], high=self.init_range[1], size=(dimensions,))
        cost = self.cost_function(weights)
        cost_curve = []
        pos_curve = []
        for epoch in range(self.max_epochs):
            old_weights = weights.copy()
            gradient = self.gradient_function(weights)
            gradient = self.clip_gradient(gradient, 1.)
            weights -= self.learning_rate * gradient
            cost = self.cost_function(weights)
            cost_curve.append(cost)
            pos_curve.append(weights.copy())
            if np.linalg.norm(weights - old_weights,1) < self.tol:
                break

        return weights, cost, 1, epoch + 1,pos_curve,cost_curve

    @staticmethod
    def clip_gradient(gradient, threshold): # to prevent vanishing/exploding gradient
        norm = np.linalg.norm(gradient)
        if norm > threshold:
            gradient = gradient * threshold / norm
        return gradient


# Example usage
if __name__ == "__main__":
    algo = GradientDescent(rastrigin_function, seed=0)
    dimensions = 2  # Set the number of dimensions for the optimization problem
    best_agent, best_cost, _, n,pos_curve,cost_curve = algo.run(dimensions)
    print("Best agent:", best_agent)
    print("Best cost:", best_cost)
    print("Number of epochs:", n)

```

```python
import autograd.numpy as np
from autograd import grad
import numpy as np_orig


class Adam:
    def __init__(self, cost_function, learning_rate=0.01, tol=1e-4, max_epochs=100, seed=10, init_range=[-5.12,5.12]):
        self.cost_function = cost_function
        self.gradient_function = grad(self.cost_function)
        self.learning_rate = learning_rate
        self.tol = tol # convergence tolerance
        self.max_epochs = max_epochs
        self.seed = seed
        self.init_range = init_range
        self.beta1 = 0.9     # Exponential decay rate for the first moment estimates
        self.beta2 = 0.999   # Exponential decay rate for the second moment estimates
        self.epsilon = 1e-8  # Small constant for numerical stability
    def run(self, dimensions):
        np_orig.random.seed(self.seed)
        weights = np_orig.random.uniform(low=self.init_range[0], high=self.init_range[1], size=(dimensions,))
        cost = self.cost_function(weights)
        # Initialize first and second moment variables
        m_t = np.zeros_like(weights)  # First moment vector (mean of gradients)
        v_t = np.zeros_like(weights)  # Second moment vector (uncentered variance of gradients)
        cost_curve = []
        pos_curve = []
        for epoch in range(self.max_epochs):
            old_weights = weights.copy()
            gradient = self.gradient_function(weights)
            gradient = self.clip_gradient(gradient, 1.)
            # Adam update rule
            m_t = self.beta1 * m_t + (1 - self.beta1) * gradient  # Update biased first moment estimate
            v_t = self.beta2 * v_t + (1 - self.beta2) * (gradient ** 2)  # Update biased second moment estimate

            # Bias correction
            m_t_hat = m_t / (1 - self.beta1 ** (epoch+1))  # Compute bias-corrected first moment estimate
            v_t_hat = v_t / (1 - self.beta2 ** (epoch+1))  # Compute bias-corrected second moment estimate

            weights -= self.learning_rate * (m_t_hat / (np.sqrt(v_t_hat) + self.epsilon))
            cost = self.cost_function(weights)
            cost_curve.append(cost)
            pos_curve.append(weights.copy())
            '''if the preference is to terminate based on cost change,
            the following condition can be amended'''
            if np.linalg.norm(weights - old_weights,1) < self.tol:
                break

        return weights, cost, 1, epoch + 1,pos_curve,cost_curve

    @staticmethod
    def clip_gradient(gradient, threshold):
        norm = np.linalg.norm(gradient)
        if norm > threshold:
            gradient = gradient * threshold / norm
        return gradient

######## Rastrigin #################
def rastrigin_function(x):
    x = np.atleast_1d(x)  # Ensure x is an array
    return 10 * len(x) + np.sum(np.array([xi**2 - 10 * np.cos(2 * np.pi * xi) for xi in x]))

# Example usage
if __name__ == "__main__":
    algo = Adam(rastrigin_function, seed=1)
    dimensions = 2  # Set the number of dimensions for the optimization problem
    best_agent, best_cost, _, n,pos_curve,cost_curve = algo.run(dimensions)
    print("Best agent:", best_agent)
    print("Best cost:", best_cost)
    print("Number of epochs:", n)
```

# FMA (Fixed Multi-Agent Adam)

```python
class MultiAgentAdam:  # Non-Adaptive MAGD
    def __init__(self, cost_function, num_agents=20, learning_rate=0.001,
                 conv_tol=1e-4, max_epochs=100, init_range=[-5.12, 5.12], seed=10):
        self.cost_function = cost_function
        self.gradient_function = grad(self.cost_function)
        self.num_agents = num_agents
        self.learning_rate = learning_rate
        self.max_epochs = max_epochs
        self.init_range = init_range
        self.conv_tol = conv_tol # convergence tolerance
        self.seed = seed
        # self.cost_history = None
        self.beta1 = 0.9     # Exponential decay rate for the first moment estimates
        self.beta2 = 0.999   # Exponential decay rate for the second moment estimates
        self.epsilon = 1e-8  # Small constant for numerical stability

    def run(self, dimensions):
        np_orig.random.seed(self.seed)
        agents = np_orig.random.uniform(low=self.init_range[0], high=self.init_range[1], size=(self.num_agents, dimensions))
        costs = np.array([self.cost_function(agent) for agent in agents])
        # self.cost_history = np.zeros((self.num_agents, self.convergence_iterations))
        best_agent = agents[np.argmin(costs)]

        # Initialize first and second moment variables
        m_t = np.zeros_like(agents)  # First moment vector (mean of gradients)
        v_t = np.zeros_like(agents)  # Second moment vector (uncentered variance of gradients)
        pos_curves = []
        cost_curve = []

        for epoch in range(self.max_epochs):
            pos_curves.append(agents.copy())
            old_best = best_agent.copy()
            gradients = np.array([self.gradient_function(agent) for agent in agents])
            gradients = np.array([self.clip_gradient(grad) for grad in gradients])

            # Adam update rule
            m_t = self.beta1 * m_t + (1 - self.beta1) * gradients  # Update biased first moment estimate
            v_t = self.beta2 * v_t + (1 - self.beta2) * (gradients ** 2)  # Update biased second moment estimate

            # Bias correction
            m_t_hat = m_t / (1 - self.beta1 ** (epoch + 1))  # Compute bias-corrected first moment estimate
            v_t_hat = v_t / (1 - self.beta2 ** (epoch + 1))  # Compute bias-corrected second moment estimate

            agents -= self.learning_rate * (m_t_hat / (np.sqrt(v_t_hat) + self.epsilon))
            costs = np.array([self.cost_function(agent) for agent in agents])


            best_agent = agents[np.argmin(costs)]
            cost_curve.append(np.min(costs))
            # if the preference is to terminate based on cost change,
            # the following condition can be amended
            if np.linalg.norm(best_agent - old_best, 1) < self.conv_tol:
                break

            # self.cost_history = np.roll(self.cost_history, -1, axis=1)
            # self.cost_history[:, -1] = costs

        best_agent_idx = np.argmin(costs)
        best_agent = agents[best_agent_idx]
        best_cost = costs[best_agent_idx]
        print("last_epoch:", epoch + 1)
        return best_agent, best_cost, len(costs), epoch + 1, pos_curves, cost_curve

    @staticmethod
    def clip_gradient(gradient, threshold=1.0):
        norm = np.linalg.norm(gradient)
        if norm > threshold:
            gradient = gradient * threshold / norm
        return gradient

######## Rastrigin #################
def rastrigin_function(x):
    x = np.atleast_1d(x)  # Ensure x is an array
    return 10 * len(x) + np.sum(np.array([xi**2 - 10 * np.cos(2 * np.pi * xi) for xi in x]))

# Run the Non-Adaptive MAGD
algo = MultiAgentAdam(rastrigin_function, num_agents=5, learning_rate=0.01, max_epochs=100, seed=0)
best_agent, best_cost, num_costs, num_epochs, pos_curves, cost_curve = algo.run(dimensions=2)
print(f"Best agent: {best_agent}, Best cost: {best_cost}, Number of agents: {num_costs}, Number of epochs: {num_epochs}")
```

# SBGD (Swarm Based GD)

```python
class SwarmBasedGradientDescent:
    def __init__(self, cost_function, num_agents=10, learning_rate=0.01, max_epochs=1000, init_range=[-1.0,1.0],
                 tolmerge=1e-3, tolres=1e-9, tolm=1e-4, seed=0, lambda_param=0.2, gamma=0.9):
        self.cost_function = cost_function
        self.num_agents = num_agents
        self.learning_rate = learning_rate
        self.max_epochs = max_epochs
        self.init_range = init_range
        self.tolmerge = tolmerge # tol_mrg
        self.tolres = tolres # tol_res
        self.tolm = tolm # tol_m
        self.lambda_param = lambda_param
        self.gamma = gamma
        self.seed = seed
        self.masses = np.ones(num_agents) / num_agents

    def run(self, dimensions):
        def compute_step_size(cost_function, gradient, x, h, lambda_param, gamma, rm):

            proposed_step = x - h * gradient
            proposed_cost = cost_function(proposed_step)
            current_cost = cost_function(x)
            condition = proposed_cost > (current_cost - lambda_param * rm * h * np.linalg.norm(gradient))
            while condition:
                h*=gamma
                proposed_step = x - h * gradient
                proposed_cost = cost_function(proposed_step)
                current_cost = cost_function(x)
                condition = proposed_cost > (current_cost - lambda_param * rm * h * np.linalg.norm(gradient))
            return h

        np_orig.random.seed(self.seed)
        agents = np_orig.random.uniform(self.init_range[0], self.init_range[1], size=(self.num_agents, dimensions))
        costs = np.array([self.cost_function(agent) for agent in agents])
        best_agent = agents[np.argmin(costs)]
        best_cost = np.min(costs)
        h_values = self.learning_rate * np.ones_like(costs)
        cost_curve = []
        pos_curve = []
        for epoch in range(self.max_epochs):
            old_best = best_agent.copy()
            F_min = np.min(costs)
            F_max = np.max(costs)

            eta = (costs - F_min) / (F_max - F_min+1e-16)
            self.masses = self.masses * (1 - eta)
            to_remove = self.masses < (1 / self.num_agents) * self.tolm
            agents = agents[~to_remove]
            costs = costs[~to_remove]
            eta = eta[~to_remove]
            h_values = h_values[~to_remove]
            self.masses = self.masses[~to_remove]
            self.num_agents -= np.sum(to_remove)

            self.masses[np.argmax(self.masses)] += np.sum(eta * self.masses)
            self.masses /= np.sum(self.masses)
            max_mass = np.max(self.masses)

            gradient_function = grad(self.cost_function)
            gradients = np.array([gradient_function(agent) for agent in agents])
            relative_mass = self.masses / max_mass
            h_values = np.array([compute_step_size(self.cost_function, gradient, agent, h, self.lambda_param, self.gamma, relative_mass[i])
                                 for i, (gradient, agent, h) in enumerate(zip(gradients, agents, h_values))])

            agents -= h_values[:, None] * gradients

            costs = np.array([self.cost_function(agent) for agent in agents])
            best_agent_index = np.argmin(costs)
            if costs[best_agent_index] < best_cost:
                best_cost = costs[best_agent_index]
                best_agent = agents[best_agent_index]

            distance_matrix = np.linalg.norm(agents[:, None, :] - agents[None, :, :], axis=-1)
            to_merge = distance_matrix < self.tolmerge

            mask = np.ones(len(agents), dtype=bool)
            merged_agents = []
            merged_costs = []
            merged_masses = []
            merged_h = []

            for i in range(len(agents)):
                if mask[i]:
                    merge_with = np.where(to_merge[i])[0]
                    new_agent = np.mean(agents[merge_with], axis=0)
                    merged_agents.append(new_agent)
                    merged_costs.append(np.mean(costs[merge_with]))
                    merged_h.append(np.mean(h_values[merge_with]))
                    merged_masses.append(np.sum(self.masses[merge_with]))
                    mask[merge_with] = False

            agents = np.array(merged_agents)
            h_values = np.array(merged_h)
            costs = np.array(merged_costs)
            self.masses = np.array(merged_masses)
            self.num_agents = len(agents)
            cost_curve.append(np_orig.min(costs))
            if np.linalg.norm(best_agent - old_best,1) < self.tolres:
                break

        return best_agent, best_cost, self.num_agents, epoch + 1,pos_curve,cost_curve
        @staticmethod
        def clip_gradient(gradient, threshold=1.0):
            norm = np.linalg.norm(gradient)
            if norm > threshold:
                gradient = gradient * threshold / norm
            return gradient
# Example usage
def rastrigin_function(x):
    x = np.atleast_1d(x)  # Ensure x is an array
    return 10 * len(x) + np.sum([xi**2 - 10 * np.cos(2 * np.pi * xi) for xi in x])

def example():
    algo = SwarmBasedGradientDescent(rastrigin_function, num_agents=20, learning_rate=0.01, max_epochs=100, seed=1, init_range=[-1.0,1.0])
    best_agent, best_cost, _, _,pos_curve,cost_curve = algo.run(dimensions=2)
    print(f"Best agent: {best_agent}, Best cost: {best_cost}")

example()
```

# MAGD (Multi-Agent GD)

```python


class MultiAgentGradientDescent: # Adaptive MAGD
    def __init__(self, cost_function, num_agents=20, learning_rate=0.001,
                 tol=1e-4, convergence_iterations=10, reduction_iters=10, conv_tol=1e-4,
                 max_epochs=100, init_range=[-5.12,5.12], seed=10):
        self.cost_function = cost_function
        self.gradient_function = grad(self.cost_function)
        self.num_agents = num_agents
        self.learning_rate = learning_rate
        self.max_epochs = max_epochs
        self.init_range = init_range
        self.convergence_tolerance = tol  # Tolerance for considering an agent as converged
        self.convergence_iterations = convergence_iterations # T_conv  # Number of iterations to check for convergence
        self.reduction_iters = reduction_iters  #T_red
        self.conv_tol = conv_tol # Tolerance for considering the whole algorithm converged
        self.seed = seed
        self.cost_history = None
        self.beta1 = 0.9     # Exponential decay rate for the first moment estimates
        self.beta2 = 0.999   # Exponential decay rate for the second moment estimates
        self.epsilon = 1e-8  # Small constant for numerical stability

    def run(self, dimensions):
        np_orig.random.seed(self.seed)
        agents = np_orig.random.uniform(low=self.init_range[0], high=self.init_range[1], size=(self.num_agents, dimensions))
        costs = np.array([self.cost_function(agent) for agent in agents])
        self.cost_history = np.zeros((self.num_agents, self.convergence_iterations))
        best_agent = agents[np.argmin(costs)]
         # Initialize first and second moment variables
        m_t = np.zeros_like(agents)  # First moment vector (mean of gradients)
        v_t = np.zeros_like(agents)  # Second moment vector (uncentered variance of gradients)
        pos_curves = []
        cost_curve = []
        for epoch in range(self.max_epochs):
            pos_curves.append(agents.copy())
            old_best = best_agent.copy()
            gradients = np.array([self.gradient_function(agent) for agent in agents])
            gradients = np.array([self.clip_gradient(grad) for grad in gradients])
            # Adam update rule
            m_t = self.beta1 * m_t + (1 - self.beta1) * gradients  # Update biased first moment estimate
            v_t = self.beta2 * v_t + (1 - self.beta2) * (gradients ** 2)  # Update biased second moment estimate

            # Bias correction
            m_t_hat = m_t / (1 - self.beta1 ** (epoch+1))  # Compute bias-corrected first moment estimate
            v_t_hat = v_t / (1 - self.beta2 ** (epoch+1))  # Compute bias-corrected second moment estimate

            agents -= self.learning_rate * (m_t_hat / (np.sqrt(v_t_hat) + self.epsilon))
            costs = np.array([self.cost_function(agent) for agent in agents])

            # convergence based reduction
            if epoch >= self.convergence_iterations and self.num_agents > 1:
                tmp = np.abs(self.cost_history[:, -self.convergence_iterations:] - costs[:, None])

                converged = np.all(tmp < self.convergence_tolerance, axis=1)
                bad_agents = costs > np.min(costs)  # Example threshold (adjust as needed)
                remove_agents = converged & bad_agents
                if np.any(remove_agents):
                    agents = agents[~remove_agents]
                    costs = costs[~remove_agents]
                    m_t = m_t[~remove_agents]
                    v_t = v_t[~remove_agents]
                    self.cost_history = self.cost_history[~remove_agents]
                    self.num_agents -= np.sum(remove_agents)
                    print("at epoch:" + str(epoch) + " remove " + str(np.sum(remove_agents)))
             # Update cost history
            self.cost_history = np.roll(self.cost_history, -1, axis=1)
            self.cost_history[:, -1] = costs

            # Performance based reduction
            if (epoch + 1) % self.reduction_iters == 0 and self.num_agents > 2:
                bad_agents = costs > np.mean(costs)  # Example threshold (adjust as needed)
                if np.any(bad_agents):
                    agents = agents[~bad_agents]
                    costs = costs[~bad_agents]
                    m_t = m_t[~bad_agents]
                    v_t = v_t[~bad_agents]
                    self.cost_history = self.cost_history[~bad_agents]
                    self.num_agents -= np.sum(bad_agents)
                    print("at epoch:" + str(epoch) + " remove " + str(np.sum(bad_agents)))

            best_agent = agents[np.argmin(costs)]
            cost_curve.append(np.min(costs))
            if np.linalg.norm(best_agent - old_best,1) < self.conv_tol:
                break


        best_agent_idx = np.argmin(costs)
        best_agent = agents[best_agent_idx]
        best_cost = costs[best_agent_idx]
        print("last_epoch:", epoch + 1)
        return best_agent, best_cost, len(costs), epoch + 1,pos_curves,cost_curve

    @staticmethod
    def clip_gradient(gradient, threshold=1.0):
        norm = np.linalg.norm(gradient)
        if norm > threshold:
            gradient = gradient * threshold / norm
        return gradient

######## Rastrigin #################
def rastrigin_function(x):
    x = np.atleast_1d(x)  # Ensure x is an array
    return 10 * len(x) + np.sum(np.array([xi**2 - 10 * np.cos(2 * np.pi * xi) for xi in x]))


algo = MultiAgentGradientDescent(rastrigin_function, num_agents=5, learning_rate=0.01, max_epochs=100, seed=0)
best_agent, best_cost, num_costs, num_epochs, pos_curves, cost_curve = algo.run(dimensions=2)
print(f"Best agent: {best_agent}, Best cost: {best_cost}, Number of agents: {num_costs}, Number of epochs: {num_epochs}")
```

```python
from sklearn.datasets import load_iris, load_wine, load_breast_cancer, load_digits, load_diabetes, fetch_california_housing, load_linnerud
from sklearn.preprocessing import StandardScaler
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score, mean_squared_error,r2_score
from sklearn.utils import shuffle

# Simple Neural Network Class Definition
class SimpleNN:
    def __init__(self, input_size, hidden_size, output_size, output_activation=None):
        self.input_size = input_size
        self.hidden_size = hidden_size
        self.output_size = output_size
        self.output_activation = output_activation
        self.initialize_weights()

    def initialize_weights(self):
        self.W1 = np_orig.random.randn(self.input_size, self.hidden_size)
        self.b1 = np_orig.zeros((1, self.hidden_size))
        self.W2 = np_orig.random.randn(self.hidden_size, self.output_size)
        self.b2 = np_orig.zeros((1, self.output_size))

    def forward(self, X):
        self.z1 = np.dot(X, self.W1) + self.b1
         # Clip the z1 to prevent overflow
        self.z1 = np.clip(self.z1, -50, 50)
        self.a1 = 1 / (1 + np.exp(-self.z1))
        self.z2 = np.dot(self.a1, self.W2) + self.b2

        if self.output_activation == 'softmax':
            exp_scores = np.exp(self.z2)
            return exp_scores / np.sum(exp_scores, axis=1, keepdims=True)
        elif self.output_activation == 'sigmoid':
            return 1 / (1 + np.exp(-self.z2))
        else:  # Linear for regression
            return self.z2

    def get_params(self):
        return np.concatenate([self.W1.ravel(), self.b1.ravel(), self.W2.ravel(), self.b2.ravel()])

    def set_params(self, params):

        W1_end = self.input_size * self.hidden_size
        b1_end = W1_end + self.hidden_size
        W2_end = b1_end + self.hidden_size * self.output_size
        b2_end = W2_end + self.output_size

        self.W1 = params[:W1_end].reshape(self.input_size, self.hidden_size)
        self.b1 = params[W1_end:b1_end].reshape(1, self.hidden_size)
        self.W2 = params[b1_end:W2_end].reshape(self.hidden_size, self.output_size)
        self.b2 = params[W2_end:b2_end].reshape(1, self.output_size)

# Custom cost function for neural networks with L2 regularization
def custom_cost_function(params, model, X, y, lambda_l2=0.001):
    model.set_params(params)
    predictions = model.forward(X)

    # Calculate the base cost
    if model.output_activation in ['softmax', 'sigmoid']:
        base_cost = -np.mean(np.sum(y * np.log(predictions), axis=1))
    else:  # Mean Squared Error for regression
        base_cost = np.mean((predictions - y) ** 2)
    weights = model.get_params()

    # L2 regularization term (sum of squared weights)
    l2_regularization = lambda_l2 * np.dot(weights,weights.T)

    # Total cost with regularization
    total_cost = base_cost + l2_regularization

    return total_cost



```

```python
# Custom  GD NN trainer Algorithm Class
class GradientDescentNN:
    def __init__(self, model, cost_function, learning_rate=0.01, tol=1e-4, max_epochs=100, seed=0,init_range=[-1.0,1.0]):
        self.model = model
        self.cost_function = custom_cost_function
        self.gradient_function = grad(self.cost_function)
        self.learning_rate = learning_rate
        self.tol = tol
        self.max_epochs = max_epochs
        self.seed = seed
        self.init_range = init_range
    def _init_params(self):
        np.random.seed(self.seed)
        params = self.model.get_params()
        total_params = params.size
        self.params_flat = np.random.uniform(low=self.init_range[0], high=self.init_range[1],
                                             size=total_params)
        self.model.set_params(self.params_flat)

    def run(self, X, y):
        np_orig.random.seed(self.seed)
        self._init_params()  # Get the initial weights
        params = self.model.get_params()
        for epoch in range(self.max_epochs):
            old_params = params.copy()
            gradient = self.gradient_function(params, self.model, X, y)
            gradient = self.clip_gradient(gradient, 1.0)
            params -= self.learning_rate * gradient
            self.model.set_params(params)  # Update the model with new weights
            cost = self.cost_function(params, self.model, X, y)

            if np.linalg.norm(params - old_params) < self.tol:
                break

        return params, cost, 1, epoch + 1

    @staticmethod
    def clip_gradient(gradient, threshold):
        norm = np.linalg.norm(gradient)
        if norm > threshold:
            gradient = gradient * threshold / norm
        return gradient


# Custom MAGD NN trainer Algorithm Class
class MultiAgentGradientDescentNN:
    def __init__(self, model, cost_function, num_agents=20, learning_rate=0.001,
                 tol=1e-4, convergence_iterations=10, reduction_iters=10, conv_tol=1e-4,
                 max_epochs=1000, init_range=[-1.0,1.0], seed=1):
        self.model = model
        self.cost_function = cost_function
        self.gradient_function = grad(self.cost_function)
        self.num_agents = num_agents
        self.learning_rate = learning_rate
        self.max_epochs = max_epochs
        self.init_range = init_range
        self.convergence_tolerance = tol
        self.convergence_iterations = convergence_iterations
        self.reduction_iters = reduction_iters
        self.conv_tol = conv_tol
        self.seed = seed
        self.cost_history = None
        self._init_agents()
        self.beta1 = 0.9     # Exponential decay rate for the first moment estimates
        self.beta2 = 0.999   # Exponential decay rate for the second moment estimates
        self.epsilon = 1e-8  # Small constant for numerical stability

    def _init_agents(self):
        np.random.seed(self.seed)
        params = self.model.get_params()
        self.params_flat = params.flatten()
        total_params = self.params_flat.size
        self.agents = np.random.uniform(low=self.init_range[0], high=self.init_range[1],
                                        size=(self.num_agents, total_params))

    def run(self, X, y):
        self._init_agents()
        costs = np.array([self.cost_function(agent,self.model, X, y) for agent in self.agents])
        self.cost_history = np.zeros((self.num_agents, self.convergence_iterations))
        best_agent = self.agents[np.argmin(costs)]
         # Initialize first and second moment variables
        m_t = np.zeros_like(self.agents)  # First moment vector (mean of gradients)
        v_t = np.zeros_like(self.agents)  # Second moment vector (uncentered variance of gradients)
        for epoch in range(self.max_epochs):
            old_cost_best = np.min(costs)
            old_best = best_agent.copy()
            gradients = np.array([self.gradient_function(agent,self.model, X, y) for agent in self.agents])
            gradients = np.array([self.clip_gradient(grad) for grad in gradients])
            # Adam update rule
            m_t = self.beta1 * m_t + (1 - self.beta1) * gradients  # Update biased first moment estimate
            v_t = self.beta2 * v_t + (1 - self.beta2) * (gradients ** 2)  # Update biased second moment estimate

            # Bias correction
            m_t_hat = m_t / (1 - self.beta1 ** (epoch+1))  # Compute bias-corrected first moment estimate
            v_t_hat = v_t / (1 - self.beta2 ** (epoch+1))  # Compute bias-corrected second moment estimate

            self.agents -= self.learning_rate * (m_t_hat / (np.sqrt(v_t_hat) + self.epsilon))
            costs = np.array([self.cost_function(agent,self.model, X, y) for agent in self.agents])
            # Co
            if epoch >= self.convergence_iterations and self.num_agents > 1:
                tmp = np.abs(self.cost_history[:, -self.convergence_iterations:] - costs[:, None])
                converged = np.all(tmp < self.convergence_tolerance, axis=1)
                bad_agents = costs > np.min(costs)
                remove_agents = converged & bad_agents
                if np.any(remove_agents):
                    self.agents = self.agents[~remove_agents]
                    costs = costs[~remove_agents]
                    m_t = m_t[~remove_agents]
                    v_t = v_t[~remove_agents]
                    self.cost_history = self.cost_history[~remove_agents]
                    self.num_agents -= np.sum(remove_agents)
                    print("at epoch:" + str(epoch) + " remove " + str(np.sum(remove_agents)))

            self.cost_history = np.roll(self.cost_history, -1, axis=1)
            self.cost_history[:, -1] = costs

            # Performance based reduction
            if (epoch + 1) % self.reduction_iters == 0 and self.num_agents > 1:
                bad_agents = costs > np.mean(costs)
                if np.any(bad_agents):
                    self.agents = self.agents[~bad_agents]
                    costs = costs[~bad_agents]
                    m_t = m_t[~bad_agents]
                    v_t = v_t[~bad_agents]
                    self.cost_history = self.cost_history[~bad_agents]
                    self.num_agents -= np.sum(bad_agents)
                    print("at epoch:" + str(epoch) + " remove " + str(np.sum(bad_agents)))

            best_agent = self.agents[np.argmin(costs)]
            if np.linalg.norm(best_agent - old_best) < self.conv_tol:
                break




            print(f"\r {np.min(costs)}", end="")
        best_agent_idx = np.argmin(costs)
        best_agent = self.agents[best_agent_idx]
        best_cost = costs[best_agent_idx]
        #print("last_epoch:", epoch + 1)
        return best_agent, best_cost, len(costs), epoch + 1

    @staticmethod
    def clip_gradient(gradient, threshold=1.0):
        norm = np.linalg.norm(gradient)
        if norm > threshold:
            gradient = gradient * threshold / norm
        return gradient

# Custom Adam NN trainer Algorithm Class
class AdamNN:
    def __init__(self, model, cost_function, learning_rate=0.01, tol=1e-4, max_epochs=100, seed=0,init_range=[-1.0,1.0]):
        self.model = model
        self.cost_function = custom_cost_function
        self.gradient_function = grad(self.cost_function)
        self.learning_rate = learning_rate
        self.tol = tol
        self.max_epochs = max_epochs
        self.seed = seed
        self.init_range = init_range
        self.beta1 = 0.9     # Exponential decay rate for the first moment estimates
        self.beta2 = 0.999   # Exponential decay rate for the second moment estimates
        self.epsilon = 1e-8  # Small constant for numerical stability
    def _init_params(self):
        np.random.seed(self.seed)
        params = self.model.get_params()
        total_params = params.size
        self.params_flat = np.random.uniform(low=self.init_range[0], high=self.init_range[1],
                                             size=total_params)
        self.model.set_params(self.params_flat)

    def run(self, X, y):
        np_orig.random.seed(self.seed)
        self._init_params()  # Get the initial weights
        params = self.model.get_params()
        # Initialize first and second moment variables
        m_t = np.zeros_like(params)  # First moment vector (mean of gradients)
        v_t = np.zeros_like(params)  # Second moment vector (uncentered variance of gradients)
        for epoch in range(self.max_epochs):
            old_params = params.copy()
            gradient = self.gradient_function(params, self.model, X, y)
            gradient = self.clip_gradient(gradient, 1.0)
            # Adam update rule
            m_t = self.beta1 * m_t + (1 - self.beta1) * gradient  # Update biased first moment estimate
            v_t = self.beta2 * v_t + (1 - self.beta2) * (gradient ** 2)  # Update biased second moment estimate

            # Bias correction
            m_t_hat = m_t / (1 - self.beta1 ** (epoch+1))  # Compute bias-corrected first moment estimate
            v_t_hat = v_t / (1 - self.beta2 ** (epoch+1))  # Compute bias-corrected second moment estimate

            params -= self.learning_rate * (m_t_hat / (np.sqrt(v_t_hat) + self.epsilon))
            self.model.set_params(params)  # Update the model with new weights
            cost = self.cost_function(params, self.model, X, y)

            if np.linalg.norm(params - old_params) < self.tol:
                break

        return params, cost, 1, epoch + 1

    @staticmethod
    def clip_gradient(gradient, threshold):
        norm = np.linalg.norm(gradient)
        if norm > threshold:
            gradient = gradient * threshold / norm
        return gradient

# Custom SBGD NN trainer Algorithm Class
class SwarmBasedGradientDescentNN:
    def __init__(self, model, cost_function, num_agents=10, learning_rate=0.01, max_epochs=1000, init_range=[-1.0, 1.0],
                 tolmerge=1e-3, tolres=1e-4, tolm=1e-4, seed=0, lambda_param=0.2, gamma=0.9):
        self.model = model
        self.cost_function = custom_cost_function
        self.num_agents = num_agents
        self.learning_rate = learning_rate
        self.max_epochs = max_epochs
        self.init_range = init_range
        self.tolmerge = tolmerge
        self.tolres = tolres
        self.tolm = tolm
        self.lambda_param = lambda_param
        self.gamma = gamma
        self.seed = seed
        self.masses = np.ones(num_agents) / num_agents

    def _init_params(self):
        np.random.seed(self.seed)
        params = self.model.get_params()
        total_params = params.size
        self.params_flat = np.random.uniform(low=self.init_range[0], high=self.init_range[1],
                                             size=total_params)
        self.model.set_params(self.params_flat)

    def run(self, X, y):
        def compute_step_size(cost_function, gradient, params, h, lambda_param, gamma, rm):
            proposed_step = params - h * gradient
            proposed_cost = cost_function(proposed_step, self.model, X, y)
            current_cost = cost_function(params, self.model, X, y)
            condition = proposed_cost > (current_cost - lambda_param * rm * h * np.linalg.norm(gradient))
            while condition:
                h *= gamma
                proposed_step = params - h * gradient
                proposed_cost = cost_function(proposed_step, self.model, X, y)
                condition = proposed_cost > (current_cost - lambda_param * rm * h * np.linalg.norm(gradient))
            return h

        np_orig.random.seed(self.seed)
        initial_params = self.model.get_params()
        dimensions = initial_params.size
        agents = np_orig.random.uniform(self.init_range[0], self.init_range[1], size=(self.num_agents, dimensions))
        costs = np.array([self.cost_function(agent, self.model, X, y) for agent in agents])
        best_agent = agents[np.argmin(costs)]
        best_cost = np.min(costs)
        h_values = self.learning_rate * np.ones_like(costs)

        for epoch in range(self.max_epochs):
            old_best = best_agent.copy()
            F_min = np.min(costs)
            F_max = np.max(costs)

            eta = (costs - F_min) / (F_max - F_min + 1e-16)
            self.masses = self.masses * (1 - eta)
            to_remove = self.masses < (1 / self.num_agents) * self.tolm
            agents = agents[~to_remove]
            costs = costs[~to_remove]
            eta = eta[~to_remove]
            h_values = h_values[~to_remove]
            self.masses = self.masses[~to_remove]
            self.num_agents -= np.sum(to_remove)

            self.masses[np.argmax(self.masses)] += np.sum(eta * self.masses)
            self.masses /= np.sum(self.masses)
            max_mass = np.max(self.masses)

            gradient_function = grad(self.cost_function)
            gradients = np.array([gradient_function(agent, self.model, X, y) for agent in agents])
            gradients = np.array([self.clip_gradient(grad) for grad in gradients])
            relative_mass = self.masses / max_mass
            h_values = np.array([compute_step_size(self.cost_function, gradient, agent, h, self.lambda_param, self.gamma, relative_mass[i])
                                 for i, (gradient, agent, h) in enumerate(zip(gradients, agents, h_values))])

            agents -= h_values[:, None] * gradients

            costs = np.array([self.cost_function(agent, self.model, X, y) for agent in agents])
            best_agent_index = np.argmin(costs)
            if costs[best_agent_index] < best_cost:
                best_cost = costs[best_agent_index]
                best_agent = agents[best_agent_index]

            distance_matrix = np.linalg.norm(agents[:, None, :] - agents[None, :, :], axis=-1)
            to_merge = distance_matrix < self.tolmerge

            mask = np.ones(len(agents), dtype=bool)
            merged_agents = []
            merged_costs = []
            merged_masses = []
            merged_h = []
            for i in range(len(agents)):
                if mask[i]:
                    merge_with = np.where(to_merge[i])[0]
                    new_agent = np.mean(agents[merge_with], axis=0)
                    merged_agents.append(new_agent)
                    merged_costs.append(np.mean(costs[merge_with]))
                    merged_h.append(np.mean(h_values[merge_with]))
                    merged_masses.append(np.sum(self.masses[merge_with]))
                    mask[merge_with] = False

            agents = np.array(merged_agents)
            h_values = np.array(merged_h)
            costs = np.array(merged_costs)
            self.masses = np.array(merged_masses)
            self.num_agents = len(agents)

            if np.linalg.norm(best_agent - old_best) < self.tolres:
                break

        # Set the best found parameters to the model
        self.model.set_params(best_agent)
        return best_agent, best_cost, self.num_agents, epoch + 1
    @staticmethod
    def clip_gradient(gradient, threshold=1.0):
        norm = np.linalg.norm(gradient)
        if norm > threshold:
            gradient = gradient * threshold / norm
        return gradient

```
