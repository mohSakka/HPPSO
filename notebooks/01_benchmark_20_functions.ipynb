# Benchmark: 20 Shifted Classical Functions

Compare **HPPSO** against PSO, PSO-m, PSO-RIW, GA-MPC, GWO, SHADE, CMA-ES, Sep-CMA-ES, and CSA on 20 benchmark functions with fixed shift vectors.

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from hppso.benchmarks.classical import build_problem_suite, FUNCTION_NAMES
from hppso.experiments.runners import DEFAULT_ALGORITHMS, run_benchmark
from hppso.utils.plotting import plot_convergence
from hppso.utils.statistics import overall_average_ranks
```

```python
DIM = 30
NUM_RUNS = 5          # increase to 20 for publication results
POP_SIZE = 30
MAX_ITERS = 200       # increase to 500 for publication results

np.random.seed(42)
problems = build_problem_suite(dim=DIM, random_shift=False)
print(f"Loaded {len(problems)} functions: {FUNCTION_NAMES[:3]} ...")
```

```python
results = run_benchmark(
    DEFAULT_ALGORITHMS,
    problems,
    num_runs=NUM_RUNS,
    pop_size=POP_SIZE,
    max_iters=MAX_ITERS,
)
```

```python
rows = []
for algo, prob_results in results.items():
    for func, data in prob_results.items():
        rows.append({
            "Function": func,
            "Algorithm": algo,
            "Mean Best Score": data["avg_final_score"],
            "Std Best Score": np.nanstd(data["all_final_scores"]),
        })

df = pd.DataFrame(rows)
display(df.pivot_table(index="Function", columns="Algorithm", values="Mean Best Score"))
print("\nOverall average ranks (lower is better):")
print(overall_average_ranks(df))
```

```python
# Plot convergence for the first function
func = FUNCTION_NAMES[0]
curves = [results[algo][func]["avg_history"] for algo in DEFAULT_ALGORITHMS]
labels = list(DEFAULT_ALGORITHMS.keys())
plot_convergence(curves, labels, f"Convergence on {func}", f"convergence_{func}")
```
