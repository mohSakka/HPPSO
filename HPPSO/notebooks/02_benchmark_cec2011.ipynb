# Benchmark: CEC2011 Real-World Problems

Requires `minionpy`: `pip install minionpy`

```python
import pandas as pd
from hppso.benchmarks.cec2011 import load_cec2011_problems
from hppso.experiments.runners import DEFAULT_ALGORITHMS, run_benchmark
```

```python
problems = load_cec2011_problems()
for p in problems:
    print(f"{p['name']} | dim={p['dimension']} | bounds={p['bounds']}")
```

```python
results = run_benchmark(
    DEFAULT_ALGORITHMS,
    problems,
    num_runs=3,
    pop_size=50,
    max_iters=100,
)
```

```python
rows = []
for algo, prob_results in results.items():
    for name, data in prob_results.items():
        rows.append({"Problem": name, "Algorithm": algo, "Mean Score": data["avg_final_score"]})
pd.DataFrame(rows).pivot_table(index="Problem", columns="Algorithm", values="Mean Score")
```
